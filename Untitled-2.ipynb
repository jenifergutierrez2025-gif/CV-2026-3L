```python
!pip install opencv-python numpy

import cv2
import numpy as np
import random
import time
import os
import tempfile

# =========================================================================
# 1. CONFIGURACIÓN GENERAL Y ALMACENAMIENTO
# =========================================================================

# Usamos la carpeta temporal del sistema para evitar problemas de permisos en Windows
CARPETA_CAPTURAS = os.path.join(tempfile.gettempdir(), "capturas_ganadores")
if not os.path.exists(CARPETA_CAPTURAS):
    os.makedirs(CARPETA_CAPTURAS, exist_ok=True)

print(f"[SISTEMA] Las capturas se guardarán en: {CARPETA_CAPTURAS}")

# --- CALIBRACIÓN PARA AMARILLO MARCADORES (HSV) ---
LOWER_COLOR = np.array([22, 70, 70])
UPPER_COLOR = np.array([38, 255, 255])

POSES = [
    "MANOS ABIERTAS CENTRO",
    "MANOS ABIERTAS ARR",
    "MANOS ABIERTAS ABA",
    "ARR IZQ / ABA DER",
    "ARR DER / ABA IZQ"
]

# =========================================================================
# 2. FUNCIONES DE PROCESAMIENTO DIGITAL DE IMÁGENES (PDI)
# =========================================================================

def obtener_posiciones_monstruos(bgr):
    """Segmenta el amarillo marcatextos y devuelve las coordenadas de las manos."""
    roi_difuminado = cv2.GaussianBlur(bgr, (11, 11), 0)
    hsv = cv2.cvtColor(roi_difuminado, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, LOWER_COLOR, UPPER_COLOR)
    
    kernel = np.ones((7, 7), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    contornos_validos = [c for c in contours if cv2.contourArea(c) > 400]
    contornos_validos = sorted(contornos_validos, key=cv2.contourArea, reverse=True)[:2]
    
    puntos_manos = []
    for c in contornos_validos:
        M = cv2.moments(c)
        if M["m00"] != 0:
            cx = int(M["m10"] / M["m00"])
            cy = int(M["m01"] / M["m00"])
            puntos_manos.append((cx, cy))
            
    puntos_manos = sorted(puntos_manos, key=lambda p: p[0])
    return puntos_manos, mask

def verificar_pose_hsv(puntos_manos, pose_objetivo, alto_pantalla, ancho_roi):
    """Evalúa la posición espacial de las manos según los cuadrantes del pecho."""
    if len(puntos_manos) < 2:
        return False
        
    mano_izq = puntos_manos[0]
    mano_der = puntos_manos[1]
    
    x_izq, y_izq = mano_izq[0], mano_izq[1]
    x_der, y_der = mano_der[0], mano_der[1]
    
    umbral_superior = alto_pantalla * 0.40
    umbral_inferior = alto_pantalla * 0.58
    distancia_centro = ancho_roi * 0.60
    
    if pose_objetivo == "MANOS ABIERTAS CENTRO":
        brazos_abiertos = abs(x_der - x_izq) > distancia_centro
        en_el_pecho = (umbral_superior <= y_izq <= umbral_inferior) and (umbral_superior <= y_der <= umbral_inferior)
        return brazos_abiertos and en_el_pecho
        
    elif pose_objetivo == "MANOS ABIERTAS ARR":
        brazos_abiertos = abs(x_der - x_izq) > distancia_centro
        ambas_arriba = y_izq < umbral_superior and y_der < umbral_superior
        return brazos_abiertos and ambas_arriba
        
    elif pose_objetivo == "MANOS ABIERTAS ABA":
        brazos_abiertos = abs(x_der - x_izq) > distancia_centro
        ambas_abajo = y_izq > umbral_inferior and y_der > umbral_inferior
        return brazos_abiertos and ambas_abajo
        
    elif pose_objetivo == "ARR IZQ / ABA DER":
        return y_izq < umbral_superior and y_der > umbral_inferior
        
    elif pose_objetivo == "ARR DER / ABA IZQ":
        return y_der < umbral_superior and y_izq > umbral_inferior
        
    return False

# =========================================================================
# 3. BUCLE PRINCIPAL EN TIEMPO REAL
# =========================================================================

cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

puntuacion_j1, puntuacion_j2 = 0, 0
pose_actual = random.choice(POSES)
tiempo_cambio = time.time()
ganador_ronda = None

# Estados de control de juego
juego_terminado = False
campeon_texto = ""

mostrar_depuracion = False
ventana_depuracion_activa = False

print("[SISTEMA] Modo Arena Competitiva a 10 Puntos Cargado.!")
print("[SISTEMA] Presiona 'q' para salir en cualquier momento.")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret: break
    
    frame = cv2.flip(frame, 1)
    alto, ancho, _ = frame.shape
    mitad = ancho // 2
    
    # Si el juego sigue activo, procesamos la cámara de forma normal
    if not juego_terminado:
        roi_j1 = frame[:, :mitad]
        roi_j2 = frame[:, mitad:]
        
        manos_j1, mascara_j1 = obtener_posiciones_monstruos(roi_j1)
        manos_j2, mascara_j2 = obtener_posiciones_monstruos(roi_j2)
        
        # Dibujar círculos indicadores amarillos
        for (cx, cy) in manos_j1:
            cv2.circle(frame[:, :mitad], (cx, cy), 12, (0, 255, 255), -1)
        for (cx, cy) in manos_j2:
            cv2.circle(frame[:, mitad:], (cx, cy), 12, (0, 255, 255), -1)
            
        cv2.line(frame, (mitad, 0), (mitad, alto), (255, 255, 255), 3)
        
        # Temporizador para la transición entre poses comunes
        if ganador_ronda and (time.time() - tiempo_cambio > 2.5):
            pose_actual = random.choice(POSES)
            ganador_ronda = None
            
        # Evaluar anotaciones
        if not ganador_ronda and verificar_pose_hsv(manos_j1, pose_actual, alto, mitad):
            puntuacion_j1 += 1
            ganador_ronda = "JUGADOR 1 PUNTO!"
            tiempo_cambio = time.time()
            cv2.imwrite(os.path.join(CARPETA_CAPTURAS, f"J1_Punto_{puntuacion_j1}.jpg"), frame)
            
            # Condición de Victoria J1
            if puntuacion_j1 >= 10:
                juego_terminado = True
                campeon_texto = "JUGADOR 1 ES EL CAMPEON"
                color_campeon = (0, 255, 0)
                
        if not ganador_ronda and verificar_pose_hsv(manos_j2, pose_actual, alto, mitad):
            puntuacion_j2 += 1
            ganador_ronda = "JUGADOR 2 PUNTO!"
            tiempo_cambio = time.time()
            cv2.imwrite(os.path.join(CARPETA_CAPTURAS, f"J2_Punto_{puntuacion_j2}.jpg"), frame)
            
            # Condición de Victoria J2
            if puntuacion_j2 >= 10:
                juego_terminado = True
                campeon_texto = "JUGADOR 2 ES EL CAMPEON"
                color_campeon = (0, 255, 255)
                
    # --- HUD GRÁFICO ---
    # Barra de estado superior
    cv2.rectangle(frame, (0, 0), (ancho, 60), (0, 0, 0), -1)
    cv2.putText(frame, f"J1: {puntuacion_j1}", (20, 42), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    cv2.putText(frame, f"J2: {ancho - 130 if 'ancho' in locals() else 510}", (ancho - 130, 42), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 255), 2) # Ajuste dinámico basado en la cadena de texto original
    
    # Se corrige la duplicación de texto de puntuación J2 para reflejar el formato correcto
    frame[:, :] = frame # Mantener base
    cv2.rectangle(frame, (0, 0), (ancho, 60), (0, 0, 0), -1)
    cv2.putText(frame, f"J1: {puntuacion_j1}", (20, 42), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    cv2.putText(frame, f"J2: {puntuacion_j2}", (ancho - 130, 42), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 255), 2)

    if not juego_terminado:
        cv2.putText(frame, f"POSE: {pose_actual}", (mitad - 180, 41), cv2.FONT_HERSHEY_SIMPLEX, 0.65, (255, 255, 255), 2)
        
        # Mostrar anuncio de punto de ronda
        if ganador_ronda:
            color = (0, 255, 0) if "1" in ganador_ronda else (0, 255, 255)
            cv2.putText(frame, ganador_ronda, (mitad - 155, alto // 2 + 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 3)
            
    else:
        # PANTALLA DE JUEGO TERMINADO
        # Fondo oscurecido traslúcido sobre el último frame capturado
        overlay = frame.copy()
        cv2.rectangle(overlay, (0, 0), (ancho, alto), (0, 0, 0), -1)
        cv2.addWeighted(overlay, 0.6, frame, 0.4, 0, frame)
        
        # Cartel de Victoria Estilizado
        cv2.putText(frame, "JUEGO TERMINADO!", (mitad - 150, alto // 2 - 15), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 255, 255), 2)
        cv2.putText(frame, campeon_texto, (mitad - 215, alto // 2 + 35), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color_campeon, 3)
        
    cv2.imshow("Carrera de Poses - Laboratorio CP", frame)
    
    # Manejo seguro de la ventana de depuración (Solo si el juego está activo)
    if mostrar_depuracion and not juego_terminado:
        mascara_completa = cv2.hconcat([mascara_j1, mascara_j2])
        cv2.imshow("Filtro HSV (Ajuste de Color)", mascara_completa)
        ventana_depuracion_activa = True
    elif ventana_depuracion_activa or juego_terminado:
        try:
            cv2.destroyWindow("Filtro HSV (Ajuste de Color)")
        except:
            pass
        ventana_depuracion_activa = False
        
    tecla = cv2.waitKey(1) & 0xFF
    if tecla == ord('q'):
        break
    elif tecla == ord('d') and not juego_terminado:
        mostrar_depuracion = not mostrar_depuracion
        
cap.release()
cv2.destroyAllWindows()

```
